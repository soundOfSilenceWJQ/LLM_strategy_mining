```python
def add_market_suffix(code: str) -> str:
    code = str(code).strip().zfill(6)
    if code[0] in ("6", "9"):
        return f"{code}.SH"
    if code[0] in ("0", "2", "3"):
        return f"{code}.SZ"
    if code[0] in ("4", "8"):
        return f"{code}.BJ"
    raise ValueError(f"无法识别交易所的代码: {code}")

def add_market_prefix(code: str) -> str:
    code = str(code).strip().zfill(6)
    if code[0] in ("6", "9"):
        return f"sh{code}"
    if code[0] in ("0", "2", "3"):
        return f"sz{code}"
    if code[0] in ("4", "8"):
        return f"bj{code}"
    raise ValueError(f"无法识别交易所的代码: {code}")

stock_symbols = [add_market_suffix(code) for code in instrument_list]
stock_symbols_prefix = [add_market_prefix(code) for code in instrument_list]
```

```python
# 我想计算因子：close/eps，分别取自data/financial_analysis/{symbol}_financial_analysis.csv和data/historical_prices/{symbol}_historical_prices.csv中的数据

financial_df = pd.DataFrame(columns=["instrument", "date", "fund_pe", "fund_pb", 'quality_gross_profit_margin', 'quality_net_profit_margin'])
for instrument in instrument_list:
    financial_analysis_file = data_dir / "financial_analysis" / f"{add_market_suffix(instrument)}_financial_analysis.csv"
    historical_prices_file = data_dir / "historical_prices" / f"{add_market_prefix(instrument)}_historical_prices.csv"
    
    if not financial_analysis_file.exists() or not historical_prices_file.exists():
        print(f"缺少数据文件: {instrument}")
        continue
    
    financial_df = pd.read_csv(financial_analysis_file)
    prices_df = pd.read_csv(historical_prices_file)
    
    # 假设financial_df中有一个列名为"EPSJB"，prices_df中有一个列名为"close"
    if "EPSJB" not in financial_df.columns or "close" not in prices_df.columns:
        print(f"数据文件格式不正确: {instrument}")
        continue
    
    calc_df = pd.DataFrame(columns=["date", "instrument", "close", "eps", "bvps", "fund_pe", "fund_pb", "quality_gross_profit_margin", "quality_net_profit_margin"])
    calc_df["date"] = pd.to_datetime(prices_df["date"])
    calc_df["instrument"] = instrument
    calc_df["close"] = prices_df["close"]
    # eps需要由公告发布日期决定，公告日之后eps才更新为最新值
    calc_df["eps"] = financial_df["EPSJB"].iloc[0]  #
    for i in range(1, len(financial_df)):
        announcement_date = pd.to_datetime(financial_df["REPORT_DATE"].iloc[i])
        last_announcement_date = pd.to_datetime(financial_df["REPORT_DATE"].iloc[i+1]) if i+1 < len(financial_df) else pd.Timestamp.max
        calc_df.loc[(calc_df["date"] < announcement_date) & (calc_df["date"] >= last_announcement_date), "eps"] = financial_df["EPSJB"].iloc[i]
        calc_df.loc[(calc_df["date"] < announcement_date) & (calc_df["date"] >= last_announcement_date), "bvps"] = financial_df["BPS"].iloc[i]
        calc_df.loc[(calc_df["date"] < announcement_date) & (calc_df["date"] >= last_announcement_date), "quality_gross_profit_margin"] = financial_df["XSMLL"].iloc[i]
        calc_df.loc[(calc_df["date"] < announcement_date) & (calc_df["date"] >= last_announcement_date), "quality_net_profit_margin"] = financial_df["XSJLL"].iloc[i]
        # 总资产周转率, roe, roa, 杠杆率
        calc_df.loc[(calc_df["date"] < announcement_date) & (calc_df["date"] >= last_announcement_date), "turnover_rate"] = financial_df["TOAZZL"].iloc[i]

    
    calc_df["fund_pe"] = calc_df["close"] / calc_df["eps"]
    calc_df["fund_pb"] = calc_df["close"] / calc_df["bvps"]
    calc_df["instrument"] = instrument
    financial_df = pd.concat([financial_df, calc_df[["instrument", "date", "fund_pe", "fund_pb", "quality_gross_profit_margin", "quality_net_profit_margin"]]], ignore_index=True)

financial_df
```

```python
import re
import numpy as np

factor_def_file = Path("factors/input/factors_financial_macro.csv")
factor_def_df = pd.read_csv(factor_def_file, engine="python", on_bad_lines="skip")
factor_def_df = factor_def_df.dropna(subset=["name", "expression"]).copy()
factor_def_df["name"] = factor_def_df["name"].astype(str).str.strip()
factor_def_df["expression"] = factor_def_df["expression"].astype(str).str.strip()
factor_def_df = factor_def_df[factor_def_df["name"].str.startswith(("fund_", "growth_", "macro_"))]

# 财务字段映射：统一到表达式中的变量名
financial_col_map = {
    "EPSJB": "eps",
    "BPS": "bvps",
    "TOTALOPERATEREVE": "revenue",
    "GROSS_PROFIT": "gross_profit",
    "PARENTNETPROFIT": "net_income",
    "PER_NETCASH": "operating_cash_flow",
    "PER_UNASSIGN_PROFIT": "retained_earnings",
}


def build_quarterly_financial_df(financial_df: pd.DataFrame) -> pd.DataFrame:
    q = financial_df.copy()
    q["date"] = pd.to_datetime(q["REPORT_DATE"], errors="coerce")
    q = q.dropna(subset=["date"]).sort_values("date").drop_duplicates(subset=["date"], keep="last")

    for raw_col, std_col in financial_col_map.items():
        if raw_col in q.columns:
            q[std_col] = pd.to_numeric(q[raw_col], errors="coerce")

    # 衍生别名，和 factors_financial_macro.csv 中表达式保持一致
    if "revenue" in q.columns:
        q["sales"] = q["revenue"]
    if "bvps" in q.columns:
        q["book_value"] = q["bvps"]
    if "operating_cash_flow" in q.columns:
        q["cash_flow"] = q["operating_cash_flow"]

    return q


macro_cache = {}


def _pick_numeric_series(df: pd.DataFrame, date_candidates: list[str], value_candidates: list[str]) -> pd.DataFrame | None:
    date_col = next((c for c in date_candidates if c in df.columns), None)
    value_col = next((c for c in value_candidates if c in df.columns), None)
    if date_col is None or value_col is None:
        return None

    out = df[[date_col, value_col]].copy()
    out.columns = ["date", "value"]
    out["date"] = pd.to_datetime(out["date"], errors="coerce")
    out["value"] = pd.to_numeric(out["value"], errors="coerce")
    out = out.dropna().sort_values("date")
    return out


def load_macro_series() -> dict[str, pd.DataFrame]:
    if macro_cache:
        return macro_cache

    # cpi: 优先用本地已下载文件，避免重复请求
    cpi_file = data_dir / "macro" / "cpi_monthly.csv"
    if cpi_file.exists():
        cpi_df = pd.read_csv(cpi_file)
        cpi_df = _pick_numeric_series(cpi_df, ["date", "日期"], ["cpi", "今值", "value"])
        if cpi_df is not None:
            cpi_df = cpi_df.rename(columns={"value": "cpi"})
            macro_cache["cpi"] = cpi_df

    # 其他宏观变量：尽量用 akshare 接口抓取，列名不一致时自动跳过
    try:
        gdp_raw = ak.macro_china_gdp()
        gdp_df = _pick_numeric_series(gdp_raw, ["季度", "日期", "date"], ["国内生产总值-绝对值", "今值", "value"])
        if gdp_df is not None:
            macro_cache["gdp"] = gdp_df.rename(columns={"value": "gdp"})
    except Exception as e:
        print(f"加载 gdp 失败: {e}")

    try:
        lpr_raw = ak.macro_china_lpr()
        lpr_df = _pick_numeric_series(lpr_raw, ["TRADE_DATE", "日期", "date"], ["LPR1Y", "1年", "value"])
        if lpr_df is not None:
            macro_cache["interest_rate"] = lpr_df.rename(columns={"value": "interest_rate"})
    except Exception as e:
        print(f"加载 interest_rate 失败: {e}")

    try:
        retail_raw = ak.macro_china_consumer_goods_retail()
        retail_df = _pick_numeric_series(retail_raw, ["日期", "date"], ["今值", "value"])
        if retail_df is not None:
            macro_cache["retail_sales"] = retail_df.rename(columns={"value": "retail_sales"})
    except Exception as e:
        print(f"加载 retail_sales 失败: {e}")

    try:
        fx_raw = ak.macro_china_fx_reserves_yearly()
        fx_df = _pick_numeric_series(fx_raw, ["日期", "date"], ["今值", "value"])
        if fx_df is not None:
            macro_cache["fx_reserves"] = fx_df.rename(columns={"value": "fx_reserves"})
    except Exception as e:
        print(f"加载 fx_reserves 失败: {e}")

    try:
        trade_raw = ak.macro_china_hgjck()
        date_col = next((c for c in ["月份", "日期", "date"] if c in trade_raw.columns), None)
        exp_col = next((c for c in ["出口金额-当月值", "出口总值", "出口", "value_export"] if c in trade_raw.columns), None)
        imp_col = next((c for c in ["进口金额-当月值", "进口总值", "进口", "value_import"] if c in trade_raw.columns), None)
        if date_col and exp_col and imp_col:
            trade_df = trade_raw[[date_col, exp_col, imp_col]].copy()
            trade_df.columns = ["date", "exports", "imports"]
            trade_df["date"] = pd.to_datetime(trade_df["date"], errors="coerce")
            trade_df["exports"] = pd.to_numeric(trade_df["exports"], errors="coerce")
            trade_df["imports"] = pd.to_numeric(trade_df["imports"], errors="coerce")
            trade_df = trade_df.dropna().sort_values("date")
            macro_cache["exports"] = trade_df[["date", "exports"]]
            macro_cache["imports"] = trade_df[["date", "imports"]]
    except Exception as e:
        print(f"加载 trade 数据失败: {e}")

    return macro_cache


def eval_expr(df: pd.DataFrame, expr: str) -> pd.Series:
    expr_py = expr
    ref_pattern = re.compile(r"Ref\(\$?([A-Za-z_][A-Za-z0-9_]*),\s*1\)")
    expr_py = ref_pattern.sub(lambda m: f"df['{m.group(1)}'].shift(1)", expr_py)

    var_pattern = re.compile(r"\$([A-Za-z_][A-Za-z0-9_]*)")
    expr_py = var_pattern.sub(lambda m: f"df['{m.group(1)}']", expr_py)

    return eval(expr_py, {"__builtins__": {}}, {"df": df, "np": np, "pd": pd})


macro_series = load_macro_series()
all_factor_rows = []
skipped_records = []

for instrument in tqdm(instrument_list, desc="计算 fund/growth/macro 因子"):
    financial_analysis_file = data_dir / "financial_analysis" / f"{add_market_suffix(instrument)}_financial_analysis.csv"
    historical_prices_file = data_dir / "historical_prices" / f"{add_market_prefix(instrument)}_historical_prices.csv"

    if not financial_analysis_file.exists() or not historical_prices_file.exists():
        skipped_records.append({"instrument": instrument, "factor": "ALL", "reason": "missing file"})
        continue

    financial_df = pd.read_csv(financial_analysis_file)
    prices_df = pd.read_csv(historical_prices_file)

    if "date" not in prices_df.columns or "close" not in prices_df.columns:
        skipped_records.append({"instrument": instrument, "factor": "ALL", "reason": "bad price schema"})
        continue

    daily = prices_df[["date", "close"]].copy()
    daily["date"] = pd.to_datetime(daily["date"], errors="coerce")
    daily["close"] = pd.to_numeric(daily["close"], errors="coerce")
    daily = daily.dropna().sort_values("date")
    daily["instrument"] = instrument

    q = build_quarterly_financial_df(financial_df)
    if q.empty:
        skipped_records.append({"instrument": instrument, "factor": "ALL", "reason": "empty financial data"})
        continue

    # 把季度财务字段映射到日频
    base_cols = [c for c in ["eps", "bvps", "revenue", "sales", "gross_profit", "net_income", "book_value", "cash_flow", "operating_cash_flow", "retained_earnings"] if c in q.columns]
    daily_fin = pd.merge_asof(
        daily.sort_values("date"),
        q[["date", *base_cols]].sort_values("date"),
        on="date",
        direction="backward",
    )

    for _, row in factor_def_df.iterrows():
        factor_name = row["name"]
        expr = row["expression"]

        try:
            if factor_name.startswith("fund_"):
                ctx = daily_fin.copy()
                # 避免除零
                for col in ctx.columns:
                    if col not in ["date", "instrument"]:
                        ctx[col] = pd.to_numeric(ctx[col], errors="coerce")
                factor_values = eval_expr(ctx, expr)
                tmp = pd.DataFrame({
                    "instrument": instrument,
                    "date": ctx["date"],
                    "factor_name": factor_name,
                    "factor": factor_values,
                })

            elif factor_name.startswith("growth_"):
                q_ctx = q[["date", *[c for c in q.columns if c != "date"]]].copy()
                factor_q = eval_expr(q_ctx, expr)
                q_factor = pd.DataFrame({"date": q_ctx["date"], "factor": factor_q}).dropna().sort_values("date")
                if q_factor.empty:
                    skipped_records.append({"instrument": instrument, "factor": factor_name, "reason": "empty growth result"})
                    continue
                mapped = pd.merge_asof(
                    daily[["date", "instrument"]].sort_values("date"),
                    q_factor,
                    on="date",
                    direction="backward",
                )
                tmp = pd.DataFrame({
                    "instrument": instrument,
                    "date": mapped["date"],
                    "factor_name": factor_name,
                    "factor": mapped["factor"],
                })

            elif factor_name.startswith("macro_"):
                # 宏观表达式在宏观时间序列上计算，再映射到个股日频
                macro_ctx = pd.DataFrame()
                # 先给每个变量补齐数据
                var_names = sorted(set(re.findall(r"\$([A-Za-z_][A-Za-z0-9_]*)", expr)))
                if not var_names:
                    skipped_records.append({"instrument": instrument, "factor": factor_name, "reason": "no macro variable"})
                    continue

                date_union = None
                for var in var_names:
                    if var not in macro_series:
                        skipped_records.append({"instrument": instrument, "factor": factor_name, "reason": f"macro var missing: {var}"})
                        macro_ctx = pd.DataFrame()
                        break
                    var_df = macro_series[var].copy()
                    if date_union is None:
                        date_union = var_df[["date"]].copy()
                    else:
                        date_union = pd.merge(date_union, var_df[["date"]], on="date", how="outer")
                if date_union is None or date_union.empty:
                    continue

                macro_ctx = date_union.sort_values("date")
                for var in var_names:
                    if var in macro_series:
                        macro_ctx = pd.merge(macro_ctx, macro_series[var], on="date", how="left")
                macro_ctx = macro_ctx.sort_values("date").ffill()

                factor_macro = eval_expr(macro_ctx.copy(), expr)
                macro_factor_df = pd.DataFrame({"date": macro_ctx["date"], "factor": factor_macro}).dropna().sort_values("date")
                if macro_factor_df.empty:
                    skipped_records.append({"instrument": instrument, "factor": factor_name, "reason": "empty macro result"})
                    continue

                mapped = pd.merge_asof(
                    daily[["date", "instrument"]].sort_values("date"),
                    macro_factor_df,
                    on="date",
                    direction="backward",
                )
                tmp = pd.DataFrame({
                    "instrument": instrument,
                    "date": mapped["date"],
                    "factor_name": factor_name,
                    "factor": mapped["factor"],
                })
            else:
                continue

            tmp = tmp.replace([np.inf, -np.inf], np.nan).dropna(subset=["factor"])
            if not tmp.empty:
                all_factor_rows.append(tmp)
            else:
                skipped_records.append({"instrument": instrument, "factor": factor_name, "reason": "all nan after calc"})

        except Exception as e:
            skipped_records.append({"instrument": instrument, "factor": factor_name, "reason": str(e)})


factor_financial_macro_df = pd.concat(all_factor_rows, ignore_index=True) if all_factor_rows else pd.DataFrame(columns=["instrument", "date", "factor_name", "factor"])

out_dir = data_dir / "quarterly"
out_dir.mkdir(parents=True, exist_ok=True)
out_file = out_dir / "factors_financial_macro_calculated.csv"
factor_financial_macro_df.to_csv(out_file, index=False, encoding="utf-8-sig")

skip_df = pd.DataFrame(skipped_records)
skip_file = out_dir / "factors_financial_macro_skipped.csv"
skip_df.to_csv(skip_file, index=False, encoding="utf-8-sig")

print(f"已生成因子记录数: {len(factor_financial_macro_df)}")
print(f"成功输出: {out_file}")
print(f"跳过记录数: {len(skip_df)}")
print(f"跳过明细: {skip_file}")
factor_financial_macro_df.head()
```
